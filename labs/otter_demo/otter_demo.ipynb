### **1.3 Functions**

Just like in normal `.py` files, you can define functions and then use them throughout your notebook.

```python
def get_greater_than(input_list, number):
    new_list = []

    for item in input_list:
        if item > number:
            new_list.append(item)

    return new_list
```

```python
item = [1,10,2,9,3,8,4,7,5,6]

get_greater_than(item, 5)
```
